```python
%%writefile app.py

from sentence_transformers import SentenceTransformer
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
from pyngrok import ngrok
import os
import pandas as pd
import streamlit as st


#Load dataset
df = pd.read_csv("/content/Food Ingredients and Recipe Dataset with Image Name Mapping.csv")

recipes = []
for _, row in df.iterrows():
    title = row['Title']
    # Assuming ingredients are comma-separated in a string
    ingredients = row['Cleaned_Ingredients'].split(',')
    recipes.append({"title": title, "ingredients": ingredients})

# Initialize the model
model = SentenceTransformer('all-MiniLM-L6-v2')

recipe_titles = [recipe['title'] for recipe in recipes]
title_embeddings = model.encode(recipe_titles)

def get_ingredients(query):
    # Encode the query
    query_embedding = model.encode([query])

    # Calculate cosine similarities
    similarities = cosine_similarity(query_embedding, title_embeddings)

    # Find the closest recipe
    closest_index = np.argmax(similarities)

    # Return the ingredients for the closest match
    return recipes[closest_index]['ingredients']

st.title("Recipe RAG with Streamlit")

query = st.text_input("Enter a Recipe search query:")
if query:
    ingredients = get_ingredients(query)
    st.write(ingredients)
```
